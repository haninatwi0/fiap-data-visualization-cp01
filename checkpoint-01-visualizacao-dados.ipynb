## 4. Visualização 1 — Relação entre peso corporal e peso total levantado

### Pergunta investigativa

Existe uma relação entre o peso corporal do atleta e o peso total levantado?

Nesta visualização, será utilizado um gráfico de dispersão para investigar a relação entre `BodyweightKg` e `TotalKg`.

### 4.1 Gráfico padrão

Primeiramente, será gerada uma versão padrão do gráfico, sem personalizações adicionais, para posteriormente compará-la com a versão customizada.

```python
sns.scatterplot(
    data=analysis_df,
    x="BodyweightKg",
    y="TotalKg"
)
```

### 4.2 Gráfico customizado

A versão customizada será desenvolvida para facilitar a identificação da relação entre o peso corporal e o peso total levantado.

```python
fig = px.scatter(
    analysis_df,
    x="BodyweightKg",
    y="TotalKg",
    color="Sex",
    opacity=0.35,
    title="Relação entre Peso Corporal e Peso Total Levantado",
    labels={
        "BodyweightKg": "Peso corporal (kg)",
        "TotalKg": "Peso total levantado (kg)",
        "Sex": "Sexo"
    }
)

fig.show()
```

### 4.3 Análise da personalização

A versão padrão permite observar a relação entre o peso corporal e o peso total levantado, porém apresenta pouca informação contextual para facilitar a comparação entre diferentes grupos.

Na versão customizada, foram adicionados título, identificação dos eixos e diferenciação por sexo. A diferenciação por sexo permite observar se a relação entre peso corporal e peso total apresenta padrões diferentes entre os grupos. A transparência dos pontos (`opacity=0.35`) também foi utilizada para reduzir a sobreposição visual e facilitar a identificação das regiões com maior concentração de observações.

Essas alterações foram escolhidas com o objetivo de facilitar a interpretação dos dados, e não apenas modificar a aparência do gráfico.


### 4.4 Interpretação dos resultados

Existe uma tendência positiva entre o peso corporal e o peso total levantado, embora exista grande dispersão dos pontos. Isso indica que atletas mais pesados tendem, em média, a levantar mais peso no total, mas o peso corporal sozinho não explica completamente o desempenho.

Também é possível observar diferenças entre os grupos de sexo. Os pontos representam uma grande quantidade de atletas, e existe uma concentração maior de observações nas faixas intermediárias de peso corporal e peso total.

Portanto, os resultados indicam que o peso corporal está associado ao desempenho total no powerlifting, mas não é o único fator relacionado ao resultado. O sexo, o equipamento utilizado, a categoria de peso e outras características dos atletas também podem influenciar o desempenho.

## 5. Visualização 2 — Equipamento e desempenho médio

### Pergunta investigativa

Como o desempenho médio varia entre os diferentes tipos de equipamento?

```python
equipment_avg = (
    analysis_df
    .groupby("Equipment", as_index=False)["TotalKg"]
    .mean()
    .sort_values("TotalKg", ascending=False)
)

equipment_avg
```

### 5.1 Gráfico padrão

Primeiramente, será apresentada a versão padrão do gráfico para comparar o desempenho médio entre os diferentes tipos de equipamento.

```python
fig = px.bar(
    equipment_avg,
    x="Equipment",
    y="TotalKg"
)

fig.show()
```

### 5.2 Gráfico customizado

Na versão customizada, as categorias serão organizadas de acordo com o desempenho médio, facilitando a comparação entre os tipos de equipamento. Os valores também serão apresentados diretamente nas barras.

```python
fig = px.bar(
    equipment_avg.sort_values("TotalKg"),
    x="TotalKg",
    y="Equipment",
    orientation="h",
    text="TotalKg",
    title="Desempenho médio por tipo de equipamento",
    labels={
        "TotalKg": "Peso total médio (kg)",
        "Equipment": "Equipamento"
    }
)

fig.update_traces(
    texttemplate="%{text:.1f} kg",
    textposition="outside"
)

fig.update_layout(
    xaxis_title="Peso total médio (kg)",
    yaxis_title="Tipo de equipamento"
)

fig.show()
```

### 5.3 Análise da personalização

A versão padrão permite comparar o peso total médio entre os diferentes tipos de equipamento, porém a posição original das categorias pode dificultar a identificação das maiores e menores médias.

Na versão customizada, as categorias são ordenadas de acordo com o peso total médio, permitindo uma comparação mais direta entre os equipamentos. Além disso, os valores são apresentados diretamente nas barras, reduzindo a necessidade de estimar os valores utilizando apenas o eixo do gráfico.

Essas personalizações foram escolhidas para destacar as diferenças entre os grupos e tornar a comparação mais objetiva. Portanto, as alterações possuem uma finalidade analítica e não apenas estética.


### 5.4 Interpretação dos resultados

O gráfico mostra diferenças no peso total médio entre os diferentes tipos de equipamento. Algumas categorias apresentam médias superiores a outras, indicando que os resultados registrados variam de acordo com o equipamento utilizado.

Essa diferença pode estar relacionada às características dos equipamentos e às condições em que as competições foram realizadas. Entretanto, não é possível afirmar que o equipamento, isoladamente, causa um aumento ou redução do desempenho, pois o resultado também pode ser influenciado por fatores como sexo, peso corporal, categoria de peso e nível dos atletas.

Assim, a visualização permite identificar uma associação entre o tipo de equipamento e o desempenho médio, mas não estabelece uma relação de causa e efeito.


## 6. Visualização 3 — Distribuição do peso total levantado

### Pergunta investigativa

Como está distribuído o peso total levantado pelos atletas?

### 6.1 Gráfico padrão

Primeiramente, será apresentada a distribuição do peso total levantado utilizando um histograma em sua configuração padrão.

```python
fig = px.histogram(
    analysis_df,
    x="TotalKg"
)

fig.show()
```

### 6.2 Gráfico customizado

Na versão customizada, foi utilizado `nbins=60` para aumentar o nível de detalhe da distribuição e permitir identificar melhor as faixas de concentração dos valores. Também foram adicionados um título e a identificação dos eixos para facilitar a compreensão do gráfico.


```python
fig = px.histogram(
    analysis_df,
    x="TotalKg",
    nbins=60,
    title="Distribuição do peso total levantado",
    labels={
        "TotalKg": "Peso total levantado (kg)"
    }
)

fig.update_layout(
    xaxis_title="Peso total levantado (kg)",
    yaxis_title="Número de atletas"
)

fig.show()
```

### 6.3 Análise da personalização

A versão padrão permite observar a distribuição geral do peso total levantado, porém a configuração padrão do histograma pode não apresentar o nível de detalhe necessário para identificar claramente as faixas de concentração dos valores.

Na versão customizada, foi utilizado `nbins=60` para aumentar o nível de detalhe da distribuição e permitir identificar melhor as faixas de concentração dos valores. Também foram adicionados um título e a identificação dos eixos para facilitar a compreensão do gráfico.

A escolha do número de intervalos foi feita com o objetivo de melhorar a leitura da distribuição e possibilitar uma análise mais clara da concentração dos dados e da presença de valores extremos.


### 6.4 Interpretação dos resultados

A distribuição mostra que a maior parte dos registros está concentrada em determinadas faixas de peso total, enquanto uma quantidade menor de atletas apresenta valores muito elevados. Isso indica que o desempenho registrado no dataset não está distribuído de maneira uniforme.

A presença de valores mais altos também demonstra que existe uma diferença considerável entre os desempenhos dos atletas. Esses valores extremos podem representar atletas com desempenho excepcional ou participantes de diferentes níveis e categorias de competição.

Portanto, o histograma permite compreender o comportamento geral do desempenho no dataset e identificar tanto as faixas mais comuns quanto a dispersão dos resultados.


## 7. Visualização 4 — Categorias de peso e desempenho

### Pergunta investigativa

Como o desempenho varia entre as categorias de peso mais frequentes?

Como o dataset possui muitas categorias de peso, foram selecionadas as oito categorias com maior número de registros. Essa escolha reduz a sobreposição visual e permite uma comparação mais clara entre grupos com maior representatividade no dataset.

```python
top_weight_classes = (
    analysis_df["WeightClassKg"]
    .value_counts()
    .head(8)
    .index
)

top_weight_classes
```

```python
weight_class_df = analysis_df[
    analysis_df["WeightClassKg"].isin(top_weight_classes)
].copy()

print("Categorias selecionadas:")
print(top_weight_classes.tolist())

print("\nNúmero de registros:")
print(len(weight_class_df))
```

### 7.1 Gráfico padrão

Primeiramente, será apresentada a distribuição do peso total para as categorias de peso mais frequentes utilizando um box plot em sua configuração padrão.

```python
fig = px.box(
    weight_class_df,
    x="WeightClassKg",
    y="TotalKg"
)

fig.show()
```

### 7.2 Gráfico customizado

Na versão customizada, as categorias serão organizadas de acordo com a mediana do peso total. Também serão adicionados título, identificação dos eixos e pontos individuais para facilitar a análise da distribuição e dos valores extremos.

```python
class_order = (
    weight_class_df
    .groupby("WeightClassKg")["TotalKg"]
    .median()
    .sort_values()
    .index
    .tolist()
)

fig = px.box(
    weight_class_df,
    x="WeightClassKg",
    y="TotalKg",
    category_orders={"WeightClassKg": class_order},
    points="outliers",
    title="Distribuição do peso total nas categorias de peso mais frequentes",
    labels={
        "WeightClassKg": "Categoria de peso (kg)",
        "TotalKg": "Peso total levantado (kg)"
    }
)

fig.update_layout(
    xaxis_title="Categoria de peso (kg)",
    yaxis_title="Peso total levantado (kg)"
)

fig.show()
```

### 7.3 Análise da personalização

O box plot padrão permite comparar a distribuição dos valores entre as categorias, mas a ordenação original das categorias não necessariamente facilita a comparação do desempenho.

Na versão customizada, as categorias são organizadas de acordo com a mediana do peso total. Dessa forma, torna-se mais fácil identificar quais categorias apresentam valores centrais maiores ou menores.

A apresentação dos outliers permite identificar desempenhos muito acima ou abaixo da distribuição central de cada categoria. Essas personalizações foram escolhidas para facilitar a comparação entre os grupos e destacar características importantes da distribuição dos dados.


### 7.4 Interpretação dos resultados

O gráfico mostra diferenças na distribuição do peso total entre as categorias de peso analisadas. As posições das medianas permitem comparar o desempenho típico de cada categoria, enquanto o tamanho das caixas indica diferenças na dispersão dos resultados.

Também é possível observar a presença de outliers em algumas categorias, representando atletas com desempenhos muito acima ou abaixo da distribuição central do respectivo grupo.

Os resultados indicam que as categorias de peso apresentam diferentes padrões de desempenho. Entretanto, essa associação não deve ser interpretada como uma relação causal isolada, pois outros fatores, como sexo, equipamento, nível competitivo e características individuais dos atletas, também podem influenciar o desempenho.
